## Dot Product Example

```python
np.random.seed(0)
sales = np.random.randint(0,20, size=(5,3))
sales
```

```python
sales_df = pd.DataFrame(sales, 
                        index=["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"], 
                        columns=["Almond Butter", "Peanut Butter", "Cashew Butter"])
sales_df
```

```python
prices = np.array([10, 8, 12])
prices.shape
```

```python
prices_df = pd.DataFrame(prices.reshape(1,3), 
                         index=["Price"], 
                         columns=["Almond Butter", "Peanut Butter", "Cashew Butter"])
prices_df
```

```python
sales_df["Total"] = np.dot(sales_df, prices_df.T)
sales_df
```
