## Matplotlib example workflow

```python
# Step 0. Import matplotlib
%matplotlib inline
import matplotlib.pyplot as plt

# Step 1. Prepare data
x = [0,1,2,3,4,5,6]
y = [40,20,10,5,10,20,40]

# Step 2. Setup plot
fig, ax = plt.subplots(figsize=(10,10))

# Step 3. Plot data
ax.plot(x,y)

# Step 4. Customize plot
ax.set(title="Simple Plot", xlabel="x-axis", ylabel="y-axis")

# Step 5. Save and show
fig.savefig("Images/sample-plt.png")
```

## Making figures from numpy arrays
Types of graphs
1. Line plot
2. Scatter plot
3. Bar plot
4. Histogram
5. Subplot

```python
import numpy as np
```

```python
# creating some data
x = np.linspace(0,10,100)
```

```python
# line plot from np array
fig, ax = plt.subplots()
ax.plot(x, np.exp(x))
```

```python
# scatter plot from np array
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
```

```python
# bar plot from a dictionary
sales = {"Almonds":10, "Cashewnut":12, "Peanuts": 5}
fig, ax = plt.subplots()
ax.bar(sales.keys(), sales.values());
ax.set(title="Nut Prices", ylabel="Price ($)", xlabel="Nut types");
```

```python
# horizontal bar plot from a dictionary
fig, ax = plt.subplots()
ax.barh(sales.keys(), sales.values());
ax.set(title="Nut Prices", ylabel="Price ($)", xlabel="Nut types");
```

```python
 # histogram from a normally distributed np array
x = np.random.normal(loc=5, scale=2, size=(1000,))
fig, ax = plt.subplots()
ax.hist(x);
```

## Subplots

```python
# option 1
x = np.linspace(0,10,100)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10,5))  # creating 4 axes

ax1.plot(x, np.sin(x))
ax2.scatter(np.random.randint(0,100, size=(10,)), np.arange(0, 10, 1))
ax3.bar(sales.keys(), sales.values())
ax4.hist(np.random.normal(loc=5, scale=0.5, size=(1000,)));
```

```python
# option 2
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,5))  # creating 4 axes

ax[0,0].plot(x, np.sin(x))
ax[0,1].scatter(np.random.randint(0,100, size=(10,)), np.arange(0, 10, 1))
ax[1,0].bar(sales.keys(), sales.values())
ax[1,1].hist(np.random.normal(loc=5, scale=0.5, size=(1000,)));
```

## Plotting Pyplot vs OO Method
* Pyplot = quick
* OO method = for advanced data

```python
#pyplot method
heart_df.plot.hist(figsize=(10,30), subplots=True);
```

```python
heart_df
```

```python
over50 = heart_df[heart_df["age"] > 50]
over50
```

```python
# OO method (sorta)
fig, ax = plt.subplots(figsize=(10,5))
over50.plot(kind="scatter", x="age", y="chol", c="target", ax=ax);
ax.set_xlim([45,100])
```

```python
# OO method (fr)
figChol, axChol = plt.subplots(figsize=(10,6))
scatter = axChol.scatter(x=over50["age"], y=over50["chol"], c=over50["target"])
axChol.set(title="Heart Disease and Cholesterolls", xlabel="Age", ylabel="Cholesterol")
axChol.legend(*scatter.legend_elements(), title="Target")
axChol.axhline(over50["chol"].mean(), linestyle="--");
```

```python
over50
```

```python
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10,10))
fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold")

# chol
scatter = ax[0].scatter(x=over50["age"], y=over50["chol"], c=over50["target"])
ax[0].set(title="Heart Disease and Cholesterol Levels", xlabel="Age", ylabel="Cholesterol")
ax[0].legend(*scatter.legend_elements(), title="Target")
ax[0].axhline(over50["chol"].mean(), linestyle="--")

#thalach
histogram = ax[1].hist(x=over50["thalach"], bins=10)
ax[1].set(title="Heart Disease and Max Heart Rate", xlabel="Max Heart Rate", ylabel="Frequency Density")
# ax[1].legend(*scatter.legend_elements(), title="Target")
# ax[1].axhline(over50["thalach"].mean(), linestyle="--")
```
