# Creating Data

## Creating a new column

```python
seats_column = pd.Series([5, 5, 5, 2, 2, 5, 5, 2, 4])
car_sales["Seats"] = seats_column
car_sales
```

```python
car_sales["Seats"] = car_sales["Seats"].fillna(int(car_sales["Seats"].mean(numeric_only=True) - 1))
car_sales
```

```python
car_sales["Fuel per 100KM"] = pd.Series([6.5, 5.8, 4.0, 10.1, 7.8, 8.8, 6.2])
car_sales
```

```python
car_sales["Fuel per 100KM"].sum(skipna=True)/10
```

```python
car_sales["Fuel per 100KM"] = car_sales["Fuel per 100KM"].fillna(car_sales["Fuel per 100KM"].sum(skipna=True)/10)
car_sales
```

```python
car_sales["Total fuel consumption (L)"] = car_sales["Odometer (KM)"]/100 * car_sales["Fuel per 100KM"]  
car_sales
```

```python
car_sales["Number of Wheels"] = pd.Series(dtype=int)
car_sales["Number of Wheels"] = car_sales["Number of Wheels"].fillna(4)
# type(car_sales["Number of Wheels"].fillna(4)) --> Series
car_sales
```

# Shuffling the data

```python
car_sales = car_sales.drop(columns=["index"])
car_sales
```

```python
car_sales = car_sales.sample(frac=1)
car_sales
```

```python
car_sales.reset_index(inplace = True, drop = True)
car_sales
```

# Applying Functions to a Series

```python
def km_to_miles(km)
    return km/1.6

car_sales["Odometer (KM)"] = car_sales["Odometer (KM)"].apply(km_to_miles)
```

```python
car_sales["Odometer (KM)"] = car_sales["Odometer (KM)"].apply(lambda x: x / 1.6)
car_sales
```

```python
car_sales.rename(columns = {"Odometer (KM)": "Odometer (Miles)"}, inplace = True)
car_sales
```
